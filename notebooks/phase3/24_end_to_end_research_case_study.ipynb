```python
from src.data_collection.pubmed_collector import collector
from src.gap_discovery.gap_detector import discoverer
from src.hypothesis_generation.generator import generator
from src.evidence_challenge.challenger import challenger
from src.refinement.hypothesis_refiner import refiner
from src.ranking.hypothesis_ranker import ranker
```

```python
theme = 'Mobile health interventions for improving adherence in type 1 diabetes'
print(f'=== Case Study: {theme} ===')
```

```python
gaps = discoverer.discover(theme)
print(f'Step 1: Top gap identified: {gaps[0]["description"][:80]}...')
```

```python
hyps = generator.generate(theme, num_hypotheses=1, mode='ontology_rag')
orig_statement = hyps[0]['statement']
print(f'Step 2: Generated Hypothesis:\n  {orig_statement}')
```

```python
challenge = challenger.challenge('case_study_1', orig_statement)
print(f'Step 3: Evidence Challenge counter-evidence count: {challenge["counter_evidence_count"]}')
```

```python
refined = refiner.refine(orig_statement, challenge)
print(f'Step 4: Refined Hypothesis Formulation:\n  {refined["refined_statement"]}')
```

```python
rank = ranker.score_hypothesis(evidence_score=0.9, novelty_score=0.85, testability_score=0.95)
print(f'Step 5: Final Composite Ranking Score: {rank}')
```
